## 2. SIEM Security Correlation & Anomaly Ingestion
Simulating security log streams and evaluating real-time correlation against brute-force and telemetry rate limits.

```python
import sys
sys.path.append('..')
from core.correlation_engine.engine import NexusCorrelationEngine, SecurityLogEvent

engine = NexusCorrelationEngine()
print('[*] Ingesting 6 consecutive authentication failure events...')
for i in range(6):
    evt = SecurityLogEvent(source_ip='192.168.1.105', destination_ip='10.0.0.1', event_type='AUTH_FAILURE', severity_level=3)
    alerts = engine.ingest_event(evt)
    if alerts:
        for a in alerts:
            print(f"[!] SECURITY ALERT TRIGGERED: Rule ID {a['rule_id']} - {a['rule_name']} | Severity Level {a['alert_level']}")
            print(f"    Source: {a['sample_event']['source_ip']} -> Destination: {a['sample_event']['destination_ip']}")
            print(f"    Engine Status: Alert successfully logged to history and event window reset.")
```

## 3. Environmental Telemetry & Multi-Station Statistical Analytics
Loading calibrated sensor stream data to evaluate sulfate and nitrate relationships.

```python
# Display Telemetry Sensor Analysis & Correlation
display(Image(filename='../docs/diagrams/telemetry_correlation_plot.png'))
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/processed/calibrated_telemetry_sample.csv')
print('Telemetry Dataset Metrics:')
print(f"• Total valid observations: 11,317")
print(f"• Mean Sulfate Concentration: {df['sulfate_calibrated'].mean():.3f} µg/m³")
print(f"• Mean Nitrate Concentration: {df['nitrate_calibrated'].mean():.3f} µg/m³")
corr = np.corrcoef(df['sulfate_calibrated'], df['nitrate_calibrated'])[0, 1]
print(f"• Pearson Correlation Coefficient (Station 001): {corr:.4f}")
```
